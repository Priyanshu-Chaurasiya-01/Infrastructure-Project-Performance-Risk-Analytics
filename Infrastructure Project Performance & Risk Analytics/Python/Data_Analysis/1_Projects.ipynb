**Feature Engineering**

```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Projects.csv"
df = pd.read_csv(file_path)
df
```

```python
df.info()
```

```python
date_col = ["Start_Date", "Planned_End_Date", "Actual_End_Date"]
for col in date_col:
    df[col] = pd.to_datetime(df[col], errors="coerce")
df.info()
```

```python
df["Cost_Overrun"] = round((df["Actual_Cost_Cr"]-df["Planned_Budget_Cr"]), 2)
df["Cost_Overrun_Status"] = np.select(
    [
        df["Cost_Overrun"] < 0,
        df["Cost_Overrun"] == 0,
        df["Cost_Overrun"] > 0
    ],
    [
        "Under Budget",
        "Exact Budget",
        "Exceeding Budget"
    ], default = "Not Available"
)
```

```python
df["Progress_Gap"] = (df["Actual_Progress_Pct"]-df["Planned_Progress_Pct"])
df["Progress_Risk"] = np.select(
    [
        df["Progress_Gap"] < -20,
        df["Progress_Gap"] < -10,
        df["Progress_Gap"] < 0
    ],
    [
        "Critical",
        "High",
        "Medium"
    ],
    default="Low"
)
```
