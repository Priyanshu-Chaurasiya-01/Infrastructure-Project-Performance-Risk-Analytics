```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Contractors.csv"
df = pd.read_csv(file_path)
df
```

```python
df.info()
```

```python
df["Performance_Category"] = np.select(
    [
        df["Performance_Rating"] >= 4.5,
        df["Performance_Rating"] >= 3.5,
        df["Performance_Rating"] >= 2.5
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Poor"
)
```

```python
df["Average_Delay_Days"]
```

```python
df["Experience_Category"] = np.select(
    [
        df["Experience_Years"] >= 20,
        df["Experience_Years"] >= 15,
        df["Experience_Years"] >= 10
    ],
    [
        "High Experienced",
        "Good Experience",
        "Average Experience"
    ],
    default="Beginner"
)
```

```python
df["Delay_Category"] = np.select(
    [
        df["Average_Delay_Days"] >= 25,
        df["Average_Delay_Days"] >= 10
    ],
    [
        "High Delay",
        "Medium Delay"
    ],
    default="Low Delay"
)
```

```python
df.to_csv("../../Data/Processed/Contractors.csv", index=False) 
```
