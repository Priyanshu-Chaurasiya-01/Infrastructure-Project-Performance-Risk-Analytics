```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Excel"/"3_ProjectMilestones.xlsx"
df = pd.read_excel(file_path)
df
```

```python
df.shape
df.info()
```

```python
df.drop_duplicates(subset=["Milestone_ID"], inplace=True)
```

```python
df["Project_ID"] = df["Project_ID"].ffill()
```

```python
date_col = ["Planned_Date", "Actual_Date"]
for col in date_col:
    df[col] = pd.to_datetime(df[col], errors="coerce")
```

```python
df.to_csv("../../Data/Cleaned/ProjectMilestones.csv", index=False)
```
