```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Excel"/"2_Contractors.xlsx"
df = pd.read_excel(file_path)
df
```

```python
df.shape
df.info()
```

```python
df.drop_duplicates(subset=["Contractor_ID"], inplace=True)
```

```python
plt.figure(figsize=(8,4))
sb.boxplot(x=df["Experience_Years"])
plt.show()
```

```python
avg_exp = df["Experience_Years"].mean().round(0)
df.loc[df["Experience_Years"]<0, "Experience_Years"] = avg_exp
```

```python
plt.figure(figsize=(8,4))
sb.boxplot(x=df["Projects_Completed"])
plt.show()
```

```python
plt.figure(figsize=(8,4))
sb.boxplot(x=df["Average_Delay_Days"])
plt.show()
```

```python
plt.figure(figsize=(8,4))
sb.boxplot(x=df["Performance_Rating"])
plt.show()
```
